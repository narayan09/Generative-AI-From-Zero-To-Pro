```python
#Check if a string is a palindrome.
l1="madam"
if l1 == l1[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
s = "madam"
rev = ""

for char in s:
    rev = char + rev

if s == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```
