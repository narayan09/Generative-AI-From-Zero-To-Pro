### What is OOP?
OOP is a way of structuring code around entities (objects) that contain:

State (data)

Behavior (methods)
Designing software that is scalable, maintainable, and change-friendly.

### Why needed?

Modularity: Code is organized into self-contained units.

Reusability: Classes can be reused across projects.

Scalability: Easier to extend and maintain large codebases.

Abstraction: Hide complex implementation details.

2️⃣ Class Methods

Belong to class, not instance.

Defined using: @classmethod

🔥 When Should You Use It? Use classmethod when:

You need alternative constructors

You modify class-level data

You want polymorphic class creation

Operates on class-level data

Often used as alternative constructors

Supports polymorphic object creation

Example: In backend systems, we use classmethods like from_env() to create objects from configuration. It allows subclass support automatically.

Think of a Class as a Blueprint
Imagine you have a blueprint for building a house.

The blueprint is the class.

The actual houses built from that blueprint are objects (or instances).

Now, there are two kinds of actions:

Instance Methods – Actions that a specific house can do.
Example: "Open the front door of house #123."
Each house has its own door, so you need to tell which house you're talking about.

Class Methods – Actions that affect the blueprint itself, not just one house.
Example: "Change the default number of windows for all future houses built from this blueprint."
Or: "Create a new house using a special type of wood (like a factory)."

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls1, data_str1):
        name, age = data_str1.split(",")
        return cls1(name, int(age))

u = User.from_string("Nik,30")
u.name
```

```python
class SET_USER:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    @classmethod
    def from_detyu(cls1,data_str1):
        name,age=data_str1.split(",")
        return cls1(name,int(age))
u =SET_USER.from_detyu("Nik,30")
u.name
```

It must return an object instance

Signature: def __new__(cls, *args, **kwargs):

Called after __new__

Must return None

### What is Encapsulation

Wrapping data and methods together.

Encapsulation is controlling how internal state is accessed and modified to protect business invariants.

Encapsulation in Python is about protecting internal state and exposing controlled interfaces. Although Python doesn’t enforce strict access modifiers like Java, we use naming conventions and properties to enforce data integrity and maintain invariants in production systems.

### 🔥 Why It Matters in Real Projects

Imagine a banking system.

If balance is public:

account.balance = -1000000

Now your system is corrupted.

Encapsulation prevents that.

### 🔹 What You Should Remember

✅ Hide internal state
✅ Use _protected and __private naming
✅ Use @property for controlled access
✅ Protect business rules (not just variables)
✅ Prevent invalid object state

### Encapsulation = Data Protection

```python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance  # protected

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Invalid deposit")
        self._balance += amount

    def get_balance(self):
        return self._balance

# Notice:

# _balance is not meant for direct modification.

# All changes go through methods.
```

```python
class BANK_AAACC:
    def __init__(self,balance):
        self._balance=balance
        
    def deposite(self,amount):
        if amount<=0:
            raise valueError("Invalid deposite")
        self._balance+=amount
    def get_balnce(self):
        return self._balance

u2=BANK_AAACC(10000)
#print(u2.get_balnce())
```

Python uses naming conventions, not strict access control.

Syntax	-------- Meaning

_name	---------Protected (by convention)
    
__name	---------Name-mangled (pseudo-private)

### What is Abstraction — 

    Hiding implementation details and showing only essential features.

Production-level definition:

    Abstraction defines contracts that separate what a component does from how it does it.
### abc → Abstract Base Classes
Abstraction means:

Show what an object can do,
Hide how it does it.

You use something without knowing internal details.

Real Life Example

When you use an ATM:

You press withdraw

You get money

You don’t know:

How the bank server works

How money is verified

How balance is updated

That’s abstraction.

    The "What vs. How" Rule: It focuses on what an object does rather than how it performs its internal operations.
    
     In Python, abstraction is achieved using the abc module (Abstract Base Classes) and the @abstractmethod decorator.
     
    Instantiation Limit: You cannot create an instance of an abstract class directly; it must be inherited by a subclass.
    
    Enforced Structure: If a subclass inherits from an abstract class, it must implement all abstract methods or it will also be considered abstract and raise a TypeError if instantiated.
    
    Concrete Methods: Abstract classes can contain both abstract methods (no code) and concrete methods (fully implemented code) to provide shared functionality to all subclasses.
    
    Key Benefits: It reduces code complexity, enhances security by hiding internal logic, and improves maintainability by allowing implementation changes without breaking the interface.

### 🔹 What You Should Remember

✅ Define common interface
✅ Use abc.ABC and @abstractmethod when needed
✅ Duck typing also achieves abstraction
✅ Depend on abstraction, not concrete class
✅ Helps in large systems & architecture

### Abstraction = Design Protection

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

dog = Dog()
cat = Cat()

print(dog.speak())
print(cat.speak())

# 🔥 What Just Happened?

# Animal defines WHAT must be done → speak()

# Dog and Cat define HOW it is done

    # animal.speak()

# User doesn’t care about internal logic.

# That is abstraction.
```

### Inheritance = Code Reuse


```python
class Animal:
    def eat(self):
        return "eating"
class Dog(Animal):
    def speek(self):
        return "BARK"

d1 = Dog()
d1.eat()
```

```python
# -------------------------------
# Base Class (Parent)
# -------------------------------

class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"My name is {self.name}"


# -------------------------------
# 1️⃣ Single Inheritance
# One child inherits from one parent
# -------------------------------

class Employee(Person):  # Employee IS-A Person
    def work(self):
        return f"{self.name} is working"


# -------------------------------
# 2️⃣ Multilevel Inheritance
# Grandparent -> Parent -> Child
# -------------------------------

class Manager(Employee):  # Manager inherits from Employee (which inherits Person)
    def manage(self):
        return f"{self.name} is managing team"


# -------------------------------
# 3️⃣ Hierarchical Inheritance
# Multiple children inherit from same parent
# -------------------------------

class Student(Person):  # Student IS-A Person
    def study(self):
        return f"{self.name} is studying"


# -------------------------------
# 4️⃣ Multiple Inheritance
# One child inherits from multiple parents
# -------------------------------

class Skills:
    def skill(self):
        return "Has technical skills"


class TechLead(Manager, Skills):
    # Inherits from:
    # Manager -> Employee -> Person  (multilevel chain)
    # AND
    # Skills (second parent)
    def lead(self):
        return f"{self.name} is leading project"


# -------------------------------
# Creating Objects
# -------------------------------

emp = Employee("Alice")
mgr = Manager("Bob")
stu = Student("Charlie")
lead = TechLead("David")

# -------------------------------
# Outputs
# -------------------------------

print(emp.introduce())   # from Person
print(emp.work())        # from Employee

print(mgr.introduce())   # from Person
print(mgr.work())        # from Employee
print(mgr.manage())      # from Manager

print(stu.introduce())   # from Person
print(stu.study())       # from Student

print(lead.introduce())  # from Person
print(lead.work())       # from Employee
print(lead.manage())     # from Manager
print(lead.skill())      # from Skills (multiple inheritance)
print(lead.lead())       # from TechLead
```

```python
# 🔥 Using super() (Very Important)

# If you want to extend parent method:

class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        parent_sound = super().speak()
        return parent_sound + " and Woof"

dog = Dog()
print(dog.speak())
```

### 💎 1️⃣ Diamond Problem

        A
       / \
      B   C
       \ /
        D

Class B inherits from A

Class C inherits from A

Class D inherits from both B and C

This shape looks like a diamond 💎
That’s why it’s called Diamond Problem.

Python solves diamond problem using Method Resolution Order (MRO), which ensures each class is called only once in a predictable order.

```python
class A:
    def show(self):
        print("Inside A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

d = D()
d.show()
```

```python
# 🧠 Why This Order?

# Because Python follows Method Resolution Order (MRO).
class A:
    def show(self):
        print("Inside A")

class B(A):
    def show(self):
        print("Inside B")
        super().show()

class C(A):
    def show(self):
        print("Inside C")
        super().show()

class D(B, C):
    def show(self):
        print("Inside D")
        super().show()

d = D()
d.show()
print(D.mro())
```

### 🔥 2️⃣ Method Resolution Order (MRO) — Simple Explanation

MRO means: The order in which Python looks for a method.

When you call: d.show()

Python searches in a specific order.

📊 D → B → C → A → object

Python checks:

Does D have show()? ✅ yes

Then super() goes to B

Then C

Then A

Then object

🧠 Very Simple Rule In multiple inheritance: Python searches:

From left to right

Then goes upward

That’s why:

class D(B, C)

B comes before C.

So Python checks B first.

### 🧠 Important Things to Remember

✅ MRO follows left-to-right order
✅ Each class appears only once in search path
✅ super() follows MRO order
✅ You can check MRO using .mro()

### What is polymorphism?

Polymorphism allows objects of different classes to respond to the same method call in different ways. In Python, it is commonly achieved using method overriding, duck typing, and operator overloading.

Polymorphism means:

Same method name, different behavior.

Break the word:

Poly = Many

Morph = Forms

So: One thing → many forms.

Overloading = multiple versions in one class → think "load more versions".

Overriding = child replaces parent's version → think "ride over the parent".

### 🔹 What You Should Remember

✅ Achieved via method overriding
✅ Achieved via duck typing
✅ Operator overloading is also polymorphism
✅ Removes long if-else chains
✅ Makes code scalable and flexible

### Polymorphism = Flexible Behavior

```python
class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

class Cow:
    def speak(self):
        return "Moo"

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.speak())

# 🧠 What happened?

# All classes have method speak()
# But each behaves differently.

# Same method name → different implementation.

# That’s polymorphism.
```

### Using Inheritance (Method Overriding)

Parent defines method

Child overrides it

Python calls correct version automatically

That is runtime polymorphism.

A child class can redefine a method from its parent class, enabling the same method name to have different behavior (e.g., overriding a draw() method in a Shape class).

Definition: A subclass provides a specific implementation of a method that is already defined in its parent class.

Purpose: To change or extend the behavior of inherited methods.

When resolved: At runtime (dynamic polymorphism).

In Python: Fully supported via inheritance.

```python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
```

### Method Overloading
Definition: Multiple methods in the same class that share the same name but have different parameters (number, type, or order).

Purpose: To perform similar actions with different inputs.

When resolved: At compile time (static polymorphism).

In Python: Not directly supported. You can mimic it using default arguments or *args

Method Overloading (Simulation): Python does not support traditional method overloading (multiple methods with the same name but different parameters). Instead, you can simulate it using default arguments, *args, or **kwargs

```python
class Calculator:
    def add(self, a, b, c=None):
        if c is None:
            return a + b
        else:
            return a + b + c
```
