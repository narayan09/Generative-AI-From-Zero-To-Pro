```python
from typing import Dict, Callable

class MCPServer:
    def __init__(self):
        self.tools: Dict[str, Callable] = {}

    def register_tool(self, name: str, func: Callable, description: str):
        self.tools[name] = {
            "function": func,
            "description": description
        }

    def list_tools(self):
        return {name: tool["description"] for name, tool in self.tools.items()}

    def call_tool(self, name: str, **kwargs):
        if name in self.tools:
            return self.tools[name]["function"](**kwargs)
        else:
            return "Tool not found"
```

```python
def get_weather(city: str):
    return f"Weather in {city} is 30°C"

def calculate(expression: str):
    return str(eval(expression))  # simple demo
```

```python
server = MCPServer()

server.register_tool(
    name="get_weather",
    func=get_weather,
    description="Get weather by city"
)

server.register_tool(
    name="calculate",
    func=calculate,
    description="Evaluate math expressions"
)
```

```python
print(server.list_tools())
```

```python
query = "What is weather in Pune?"

# Simulated decision
if "weather" in query.lower():
    result = server.call_tool("get_weather", city="Pune")
elif "calculate" in query.lower():
    result = server.call_tool("calculate", expression="2+2")

print("Tool Result:", result)
```
