```python
#anagram check
s1=input("input Enter a no").lower()
s2=input("input Enter a no").lower()
if len(s1) != len(s2):
    print("not anagram")
else:
    count = {}
    for c in s1:
        count[c]=c.get(c,0)+1
    for c in s2:
        if c not in count or count[c]==0:
            print("not anagram")
            break
        count[c] -= 1
    else:
        print("anagram")
```

```python
#fibbocanni series
n=int(input("Enter no of terms: "))
a,b=0,1
for i in range(n):
    print(a,end="")
    a,b=b,a+b


def fib(n):
    if n<=1:
        return n
    return fib(n-1) +fib(n-2)

n=int(input("Enter no of terms: "))
for i in range(n):
    print(fib(i),end= "")
    
```

```python
#Prime number
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage
num = int(input("Enter a number: "))

if is_prime(num):
    print(f"{num} is a Prime Number")
else:
    print(f"{num} is NOT a Prime Number")
     
```

```python
#Remove duplicates from list

def remove_duplicates(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result

# Example
data = [1, 2, 2, 3, 4, 4, 5]
print(remove_duplicates(data))
```

```python
#Second largest element
def second_largest(lst):
    unique = list(set(lst))
    unique.remove(max(unique))
    return max(unique)

# Example
data = [10, 20, 4, 45, 99]
print(second_largest(data))
```
