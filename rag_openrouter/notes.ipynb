# Generative AI

Before understanding Generative AI, let's first understand what Artificial Intelligence means.

Artificial Intelligence (AI) is the ability of computers or machines to perform tasks that normally require human intelligence, such as:

- Understanding language
- Recognizing images
- Making decisions
- Solving problems

Traditionally, AI systems were designed mainly to analyze and classify information.

For example:

- Is this email spam or not?
- Is this image a cat or a dog?
- Will a customer buy a product or not?

These systems only analyze existing data and provide predictions.

# What Makes Generative AI Different?

Generative AI goes one step further.

Instead of only analyzing data, it can create new content.

- Documents

## Definition

**Generative AI is a branch of Artificial Intelligence that creates new content based on patterns learned from large amounts of data.**

### What are LLMs? 🤖

#### 1. Definition & Analogy 🧠
Large Language Models (**LLMs**) are advanced AI systems trained to understand and generate text. 

Think of an LLM as a highly sophisticated **predictive text engine** 📱, much like the autocomplete feature on your smartphone, but on a massive scale. If you type *"A dime a..."*, your phone guesses *"dozen"*. LLMs do the exact same thing, but they can predict the next word so accurately that they can write essays, code, and solve complex problems. 

They don't "think" or "know" things the way humans do. Instead, they look at the words you give them and calculate the most statistically likely words to follow based on billions of examples they studied during training.🌐

#### 2. Key Terms 🔑
* **Training Data:** 📚 The massive collection of text (books, articles, websites) that the AI reads to learn language patterns.
* **Parameters:** 🎛️ The internal "dials" or "connections" the model adjusts during training. More parameters generally mean a more capable model.
* **Prompt:** 💬 The input text or instruction you give to the LLM.
* **Completion:** ✍️ The output text generated by the LLM in response to your prompt.

```python
import random

# A tiny sample of "training data"
training_text = "ai is amazing ai is fast ai is changing the world"
words = training_text.split()

# Create a simple lookup dictionary of what word follows what
word_pairs = {}
for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i+1]
    if current_word not in word_pairs:
        word_pairs[current_word] = []
    word_pairs[current_word].append(next_word)

# Predict the next word after "ai" based on probabilities
prompt = "ai"
if prompt in word_pairs:
    predictions = word_pairs[prompt]
    chosen_next_word = random.choice(predictions)
    print(f"Prompt: '{prompt}'")
    print(f"Predicted next word: '{chosen_next_word}'")
    print(f"All possible options learned: {predictions}")
```

### Limitations of LLMs 🚧

#### 1. Definition & Analogy 🎭
While LLMs seem incredibly smart, they have strict operational boundaries. They lack true understanding, common sense, and consciousness.

Think of an LLM as a **world-class actor reading a script** 📜. The actor can play a brilliant doctor and speak perfect medical jargon, but they cannot actually perform surgery. LLMs are fluent in the *structure* of language, but they do not possess a deeper, sentient understanding of the concepts they are typing.🧱

#### 2. Key Terms 🔑
* **No Intentionality:** 🤖 LLMs do not have goals, beliefs, or feelings; they only respond to immediate stimuli (prompts).
* **Context Window:** 🪟 The maximum amount of text (both prompt and response) that the model can process at one single time. If a conversation gets too long, it "forgets" the beginning.
* **Compute Cost:** ⚡ The massive amount of computer processing power and electricity required to train and run these models.

```python
# Simulating a model with a strict 3-word memory limit
memory_limit = 3
conversation_history = ["Hello", "I", "love", "teaching", "artificial", "intelligence"]

print("Full conversation history:", conversation_history)

# The model can only look at the last 3 words
visible_context = conversation_history[-memory_limit:]
print(f"\nModel's limited view (Context Window): {visible_context}")
print(f"The model has completely forgotten: {conversation_history[:-memory_limit]}")
```

### Hallucinations 🍄

#### 1. Definition & Analogy 🃏
A **hallucination** occurs when an LLM confidently generates information that is factually incorrect, completely fabricated, or ungrounded in reality.

Think of an LLM as an **overconfident, people-pleasing intern** 🧑‍💼. If you ask this intern a question they don't know the answer to, instead of admitting they are clueless, they will invent a highly believable, articulate story just to satisfy you. Because LLMs prioritize matching language patterns over verifying facts, incorrect answers can sound completely convincing.📣

#### 2. Key Terms 🔑
* **Factuality:** 🎯 The accuracy of the AI's output compared to real-world truths.
* **Groundedness:** ⚓ Whether the AI's response is strictly based on the provided source text or if it is guessing.
* **Temperature:** 🌡️ A setting that controls the randomness of the output. Higher temperature means more creative (and potentially more hallucinated) text.

```python
# Simulating how an LLM fabricates facts based on pattern matching
def generate_biography(name):
    # The model knows the pattern of a biography, but doesn't know this person
    template = f"{name} was born in 1945. They invented the first automated bread-slicing laser."
    return template

# Let's pass a completely made-up name
fake_name = "Professor John Doe-Smith"
print("Prompt: Write a short biography of Professor John Doe-Smith.")
print("\nLLM Response (Hallucination):")
print(generate_biography(fake_name))
print("\nNotice how confident the response sounds despite being entirely made up!")
```

### Knowledge Cutoff 📅

#### 1. Definition & Analogy 📆
An LLM's **knowledge cutoff** is the fixed date when its training process ended. The model knows absolutely nothing about world events, discoveries, or data created after this specific milestone.

Think of an LLM as a **textbook printed in a specific year** 📚. If you buy a history textbook printed in 2020, it will contain beautifully detailed chapters on everything up to that year. However, no matter how hard you look, you cannot use that book to find out who won the world cup in 2024. The book isn't broken; its information is simply frozen in time.⏳

#### 2. Key Terms 🔑
* **Static Knowledge:** 💾 Information that remains fixed and does not automatically update after training is complete.
* **Real-time Data:** ⚡ Live, current information happening right now (weather, stocks, today's news).
* **RAG (Retrieval-Augmented Generation):** 🔍 A technique where an LLM is connected to an external search engine to bypass its cutoff date.

```python
import datetime

# The date the model finished its training
KNOWLEDGE_CUTOFF_YEAR = 2025

def ask_ai_about_event(event_year):
    print(f"User Query: What happened in the year {event_year}?")
    
    if event_year <= KNOWLEDGE_CUTOFF_YEAR:
        print("LLM Response: I have data on this! Here is what happened... [Displays historical info]")
    else:
        print("LLM Response: I apologize, but my knowledge cutoff is in 2025. I do not have access to real-time data or events beyond that date.")

# Test 1: Before cutoff
ask_ai_about_event(2023)
print("-" * 40)
# Test 2: After cutoff
ask_ai_about_event(2026)
```
