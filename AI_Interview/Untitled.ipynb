# Transformer architecture (high level)

# Step 1: Why Transformer Architecture Exists

    Before Transformers, NLP models used:
    
        RNN (Recurrent Neural Network) 1980s -  A neural network that remembers past information to process sequences like sentences or time series.
        
        How it works: As you read a sentence, you remember the previous words to understand the current one. If you read "The sky is," you know the next word is probably "blue" because you remember the context.

        The Problem: If the book is 500 pages long, by the time you reach the end, you have probably forgotten a small detail mentioned on page 10.
        
        LSTM (Long Short Term Memory) 1997 - An advanced RNN that can learn long-term patterns by controlling what to remember or forget.
        How it works: While watching a movie, you jot down important clues (the killer’s hat) in a notebook and ignore irrelevant details (what the waiter looked like). When the mystery is solved at the end, you check your notebook to remember the clue from the beginning.

    The Advantage: Unlike the person reading the book (RNN), you don’t forget the clue from the first scene because your "notebook" (the LSTM) decided it was important enough to save.

Problem with them:

Example sentence:

    "The animal didn't cross the street because it was tired."
    
    To understand “it”, the model must remember animal.

    RNNs process words one by one, so:
    
    Word1 → Word2 → Word3 → Word4 → ...

    Problems:
    
    Very slow
    
    Hard to remember long sentences
    
    Cannot process words in parallel

So in 2017, Google introduced the paper:

📄 “Attention Is All You Need”

This introduced the Transformer architecture.

Transformers solve the problem using Attention.
        

# Step 2: Simple Idea of Transformer

### The key idea:

    - Instead of reading words one by one,

    - The model looks at all words together and decides:

    - Which words are important for understanding another word.

```
  Example sentence:
    The cat sat on the mat
```
If the model processes “sat”, it looks at:

- cat
- mat
- the

and decides which words matter.

This process is called:

### Self Attention

```
Self-Attention is a mechanism where a word looks at all other words in a sentence to decide which ones are most relevant to understanding its own meaning.
```

```
Imagine you are at a party trying to understand the sentence: "The animal didn't cross the street because it was too tired. "

To understand what "it" refers to, you don't just look at the word "it." You scan the whole sentence. You look at "animal" and "street." You realize that "animals" get tired, but "streets" don't. So, you decide that "it" should pay attention to the word "animal."

Self-attention is the process of the word "it" asking: "Which words in this sentence are important for me to figure out what I mean?"

Example 1: The "Bank" Sentence (Disambiguation)
Sentence: "The farmer went to the bank of the river."

The Challenge: The word "bank" is ambiguous. It could mean a financial institution or the side of a river.

How Self-Attention Works:

The word "bank" looks at the word "river."
It calculates that "river" is very relevant (high attention score).
It ignores the word "farmer" (low score) in this context.
Result: Because of the connection to "river," the model understands that "bank" here means the riverbank, not a money bank.
```

### Step 3: High Level Transformer Architecture

- The Transformer has two main parts:

```
Input Sentence
      │
      ▼
   Encoder  (The Listener)
      │
      ▼
   Decoder (The Speaker)
      │
      ▼
Output Sentence

Encoder-Only (BERT): Great for tasks where you need to understand a sentence (e.g., Is this review positive or negative?).

Decoder-Only (GPT): Great for tasks where you need to generate text (e.g., Write a poem about cats).
```

# Step 4: Transformer Encoder Structure
Each encoder block contains:

```
Input 

 │
 ▼
Embedding


Computers don't understand words; they understand numbers. This step turns every word (like "Apple") into a long list of numbers called a vector. Words with similar meanings end up having similar numbers.

 │
 ▼
Positional Encoding

Since the Transformer looks at the whole sentence at once, it doesn't naturally know the order of words. We add a mathematical "timestamp" to each word's vector so the model knows that "The dog bit the man" is different from "The man bit the dog."

 │
 ▼
Multi Head Attention

This is the "brain power" of the model. It allows every word to look at every other word to find meaning.
Simple Example: In the sentence "The bank was closed because it was a holiday," the word "it" uses attention to realize it refers to the "bank," not the "holiday."
"Multi-Head" just means it does this multiple times simultaneously to catch different patterns (like grammar vs. meaning).


 │
 ▼
Add & Normalize

Think of this as a quality control step.
Add: We take the original information from the start and add it back to the result of the attention. This ensures no original data is "lost."
Normalize: We adjust the numbers so they stay within a healthy range, preventing the model from getting "confused" by numbers that are too big or too small.


 │
 ▼
Feed Forward Network

Now that the model knows how words relate to each other (context), it processes each word individually to think about it more deeply. It’s like a mini-brain for each word that says, "Based on the context I just learned, what does this word actually represent here?"

 │
 ▼
Add & Normalize
The final quality check. It takes the deep thoughts from the Feed-Forward step, adds back the previous data, and cleans up the numbers one last time before passing the information to the next layer.

This block repeats N times (usually 6 or more).

```

### Step 5: Word Embeddings

Computers cannot understand words.

So words are converted into vectors (numbers).

Example:

- cat → [0.21, 0.87, 0.44]
- dog → [0.19, 0.82, 0.41]
- car → [0.65, 0.12, 0.77]

```Words with similar meaning have similar vectors.
This is called:
```

### Embedding
How Input Embeddings Work

The process follows three simple phases:
- Tokenization: The sentence is broken into smaller chunks called "tokens." These can be whole words or sub-words (e.g., "playing" might become "play" and "##ing").
- Integer Mapping: Each unique token in the model's vocabulary is assigned a specific ID number. For example, "the" might be 101, and "bank" might be 2045.
- Vector Lookup: This is the actual "embedding" part. The model has a giant lookup table (a matrix) where every ID number corresponds to a long list of numbers called a dense vector. 


# Step 6: Positional Encoding

Transformers read all words at the same time, so they don't know order.

```
Example:
    Dog bites man
    Man bites dog
    Same words → different meaning.
So we add position information.
Example:
```
- Dog → position 1
- bites → position 2
- man → position 3

This is called:

### Positional Encoding

It tells the model:

Which word comes first, second, third.

```
How it works (The Simple Math Secret)
The model uses a specific mathematical pattern (called Sine and Cosine waves) to create these position signals.
Why waves? Imagine a rhythmic beat. Even if the sentence is 10 words or 1,000 words long, the "wave" pattern allows the model to calculate the distance between words easily.
The Best Part: We don't just put "1, 2, 3" because those numbers get too big for the computer to handle well. Instead, we use these wave patterns to keep the numbers small and consistent.
The Final Result:

Word Meaning + Position Signal = Context-Ready Word

Now the model knows: "This is the word 'Dog', and it is the first word in the sentence."
```

# Step 8: Query, Key, Value (Internal Working)

Inside attention, each word is converted into 3 vectors:

- Query (Q)
- Key   (K)
- Value (V)

Think like search in a database.
Example:

- Query → What I am looking for
- Key → What words contain
- Value → Information of that word

Query, Key, and Value are a clever information retrieval system where a word asks a question (Query), checks against all words (Keys), and gathers the relevant information (Values).

Attention Score:
```
Attention = Q × K
Then apply: Softmax

Then multiply with: Value

Formula (interview important):

Attention(Q,K,V) = softmax(QKᵀ / √d) V
```

# Step 9: Multi Head Attention

Instead of one attention, the model uses multiple attentions simultaneously.

Example:

- Head 1 → grammar relation
- Head 2 → semantic relation
- Head 3 → object relation
- Head 4 → action relation

Then combine them. Why?

Because language relationships are complex.

### Why is this better?
By having 8 or 12 "Heads" working simultaneously, the model gets a 3D understanding of the sentence. It doesn't just read the words; it understands the structure, the logic, and the context all in one go.

Summary of the Flow:

- Split: Break the sentence into multiple "Heads."
- Search: Each Head uses Q, K, and V to find connections.
- Combine: Stitch all the experts' findings back together into one big, smart vector.

# Step 10: Feed Forward Network

The Feed-Forward Network is a simple, two-step processing layer where each word independently "thinks" about what it just learned from the team huddle.

After attention, data goes into a simple neural network.

Structure:
```
Linear Layer
     ↓
Activation (ReLU)  - It's a filter. It removes negative or useless information between the two steps. It asks, "Is this detail actually useful?" If not, it zeros it out.
     ↓
Linear Layer 
```

Purpose:

Learn deeper patterns

Transform representations

# Step 11: Add & Normalize

acts like a "safety check" for a neural network. Imagine a long assembly line; by the time a product reaches the end, mistakes can pile up. These two techniques prevent that.

The Residual Connection makes sure the AI doesn't forget what it started with, and Normalization keeps all its internal numbers from getting out of control.

Two techniques used:

- 1️⃣ Residual Connection

    Instead of replacing input:
    
    Output = Layer(x) + x
    
    Benefits:
    
    Prevents gradient problems
    
    Stabilizes training

- 2️⃣ Layer Normalization

    Normalizes values so training is stable.

# 1️⃣ Temperature (Creativity Control)

### Simple Definition

- Temperature controls how random or creative the model’s response is.

- Think of it like a creativity knob.

- Low Temperature  → Predictable answers
- High Temperature → Creative answers
```
Example Prompt:
Tell me a color of the sky.
Possible next tokens with probabilities:

Word	Probability
blue	0.7
white	0.15
grey	0.10
purple	0.05
Low Temperature (0.1)

Model picks most probable token.
Output: The sky is blue.
Very deterministic.

Medium Temperature (0.7)
Model allows some randomness.
Output could be:
The sky is blue with white clouds.

High Temperature (1.2)
More randomness.
Output:
The sky glows purple during magical sunsets.
Creative but sometimes less accurate.
```

### Typical Temperature Values

| Temperature | Behavior        |
|-------------|-----------------|
| 0           | deterministic   |
| 0.2         | factual answers |
| 0.5         | balanced        |
| 0.7         | normal chat     |
| 1.0         | creative        |
| 1           | very random     |


When to Use

Low temperature:
- Code generation
- Math
- Facts
- SQL

Higher temperature:

- Story writing
- Brainstorming
- Marketing
- Creative content
  
### 2️⃣ Top_p (Nucleus Sampling)

- Temperature controls randomness.

- Top_p controls how many possible words the model considers.

- It selects tokens whose combined probability ≤ p.

# Example

## Possible tokens

| Word  | Probability |
|-------|-------------|
| blue  | 0.5         |
| white | 0.2         |
| grey  | 0.15        |
| red   | 0.1         |
| green | 0.05        |

**Top_p = 0.6**

Model selects tokens until probability reaches 0.6:

- blue (0.5)  
- white (0.2)  

Total = 0.7 → stop  

So model only chooses between:  
**blue, white**

---

**Top_p = 0.9**

Now it includes more options:

- blue  
- white  
- grey  
- red  

More variety.

---

## Easy Way to Remember

**Temperature:**  
- controls randomness  

**Top_p:**  
- controls token pool size  

---

## Temperature vs Top_p

| Feature | Temperature | Top_p |
|---------|-------------|-------|
| Controls | randomness | candidate token pool |
| Range    | 0–2        | 0–1   |
| Effect   | probability scaling | probability filtering |

---

## Rule (Important)

Usually adjust **one only**.

**Typical settings:**
- temperature = 0.7  
- top_p = 0.9

# 3️⃣ Hallucination

## Definition
Hallucination = when an LLM generates incorrect or made-up information but presents it as true.  
The model sounds confident but is wrong.

---

## Example

**Prompt:**  
Who invented Python in 1995?

**Correct answer:**  
Guido van Rossum  

**Hallucinated answer:**  
Python was invented by James Gosling in 1995.  

This sounds believable but is false.

---

## Why Hallucination Happens
LLMs do next-token prediction, not true reasoning.  
The model predicts: **most probable text** → not necessarily true facts.

---

## Causes
1️⃣ Missing knowledge  
2️⃣ Ambiguous prompt  
3️⃣ No context  
4️⃣ High temperature  
5️⃣ Weak retrieval in RAG  

---

## Example in RAG
If RAG retrieval fails → LLM invents answer.

**Example:**  
User asks: *What is the refund policy in the document?*  
If document chunk is missing → Model guesses policy.  
This is hallucination.

---

## How to Reduce Hallucination
Methods used in production systems:

1️⃣ **RAG (Retrieval Augmented Generation)**  
Provide real data.  
Vector DB → retrieve → LLM  

2️⃣ **Lower temperature**  
Example: `temperature = 0.2`  

3️⃣ **Strong prompting**  
Example: *If you don't know the answer, say "I don't know".*  

4️⃣ **Grounding with sources**  
Example output: Answer + document citation  

5️⃣ **Structured output**  
Example: Return JSON  

---

# 1️⃣ Why Do We Need Embeddings?

Computers **cannot understand text directly**.  
Example sentence: I love pizza


Computers only understand **numbers**, not words.  
Naive mapping:
- So we must convert text → numbers.
- One simple way is assigning numbers:
```
I = 1
love = 2
pizza = 3

But this approach has a big problem.

It does not capture meaning.
Example:
pizza
burger
pasta
These are similar foods, but simple numbering does not show that.
So we need something smarter.

This is where Embeddings come in.
```



# 2️⃣ What is an Embedding?

An **embedding is a vector (list of numbers) that represents the meaning of text**.

Example: 
- cat → [0.21, 0.87, 0.44]
- dog → [0.19, 0.82, 0.41]
- car → [0.65, 0.12, 0.77]


- `cat ≈ dog` → vectors are close  
- `cat ≠ car` → vectors are far  

Embeddings capture **semantic meaning**.

```

# 5️⃣ Why Embeddings Are Used

Example query: "Best laptop for programming" -> "Top computers for software developers"

Applications:

- Search engines  
- RAG systems

- Chatbots  
- Similarity search

---

# 7️⃣ How Similarity is Calculated

### Cosine Similarity
- Measures **angle between vectors**  
- `1 → identical`, `0 → unrelated`, `-1 → opposite`  
- Most common in AI

### Euclidean Distance
- Measures **straight-line distance**  
- Short distance → similar

### Dot Product
- Used in some vector DBs  
- Faster for large-scale search

---

# 8️⃣ Evolution of Embedding Techniques

### Technique 1 — Bag of Words (Old)
- Word counts only  
- ❌ No meaning, sparse vectors

### Technique 2 — Word2Vec (Google, 2013)
- Context-based similarity  
- Famous: `king - man + woman = queen`  
- Methods: CBOW, Skip-Gram

### Technique 3 — GloVe (Stanford)
- Global co-occurrence statistics  
- Example: `doctor, hospital, patient`

### Technique 4 — FastText (Facebook)
- Handles **subwords**  
- Example: `playing, played, player` → root **play**

### Technique 5 — Transformer Embeddings (Modern)
- Contextual embeddings (BERT, OpenAI, Sentence Transformers, E5, Instructor)  
- Example:  
  - `bank` in *river bank* ≠ `bank` in *bank account*

# Prompt Engineering Techniques

Prompt engineering means **designing prompts (instructions) to guide an LLM to produce better and more accurate outputs**.

LLMs respond based on **how clearly the prompt is written**.

Two basic and very common techniques are:

- Zero-shot prompting
- Few-shot prompting

---

**Zero-shot prompting means asking the model to perform a task without giving any examples.**

The model relies only on its **pre-trained knowledge**.

Instruction + Input
Example:
```

"I love programming"

Prompt

```python

# Prompt Engineering Techniques

## 1. Chain-of-Thought (CoT) Prompting

### One Line Definition
**Chain-of-Thought prompting encourages the model to solve a problem by showing or generating step-by-step reasoning.**

Instead of jumping directly to the answer, the model explains **how it reached the answer**.

---

## Why Chain-of-Thought is Used

It helps with:

- Math problems
- Logical reasoning
- Multi-step tasks
- Complex decision making

The model performs better when it **breaks the problem into steps**.

---

## Basic Idea

Normal Prompt

```

What is 25 × 12?

```

Output

```

300

```

Chain-of-Thought Prompt

```

What is 25 × 12?
Explain step by step.

```

Output

```

25 × 12

Step 1: 25 × 10 = 250
Step 2: 25 × 2 = 50
Step 3: 250 + 50 = 300

Final Answer: 300

```

---

## Example 1 — Math Problem

Prompt

```

If a store sells 15 apples per day, how many apples will it sell in 7 days?
Let's think step by step.

```

Output

```

Step 1: Apples sold per day = 15
Step 2: Number of days = 7
Step 3: Total apples = 15 × 7 = 105

Answer: 105 apples

```

---

## Example 2 — Logical Reasoning

Prompt

```

John has 3 boxes. Each box contains 4 books.
How many books does John have?
Let's think step by step.

```

Output

```

Step 1: Books per box = 4
Step 2: Number of boxes = 3
Step 3: Total books = 3 × 4

Answer: 12 books

```

---

## Example 3 — Programming

Prompt

```

Write Python code to check if a number is even or odd.
Explain the logic step by step.

```

Output

```

Step 1: Take input number
Step 2: Use modulo operator
Step 3: If number % 2 == 0 → Even
Step 4: Else → Odd

```

---

## Simple Trick to Trigger Chain-of-Thought

Just add phrases like:

```

Let's think step by step
Explain the reasoning
Break the problem into steps
Show your reasoning

```

These often activate reasoning in LLMs.

---

## When to Use Chain-of-Thought

Use it when the task requires:

- reasoning
- calculation
- planning
- complex decisions

Examples:

- math problems
- debugging code
- data analysis
- logic puzzles

- safety rules
- response format
- domain expertise

Think of it like **setting the personality of the AI**.

---

```

Model Output → Response following system rules

# Prompt Design Principles

Prompt design principles help create **effective prompts that produce accurate, consistent, and safe responses from LLMs**.

Good prompts reduce:

- hallucination
- ambiguity
- inconsistent outputs

Four important principles:

1. Clear Instructions  
2. Output Format  
3. Context Injection  
4. Guardrails  

---

# 1. Clear Instructions

## One Line Definition
**Clear instructions mean writing prompts that explicitly tell the model what task to perform.**

The clearer the instruction, the better the output.

---

## Bad Prompt

```

Tell me about Python.

```

Problem:

- Too vague
- Model doesn't know what to focus on

---

## Better Prompt

```

Explain Python decorators in simple terms with a code example.

```

Now the model knows:

- topic
- explanation level
- example required

---

## Example 1 — Summarization

Bad Prompt

```

Explain this.

```

Better Prompt

```

Summarize the following paragraph in 3 bullet points.

```

---

## Example 2 — Coding

Bad Prompt

```

Write code.

```

Better Prompt

```

Write a Python function to check if a number is prime.
Include comments explaining each step.

```

---

## Example 3 — Classification

Bad Prompt

```

Analyze this sentence.

```

Better Prompt

```

Classify the sentiment of the sentence as Positive, Negative, or Neutral.

```

---

## Key Rule

```

Clear prompts = Better answers

```

Always specify:

- task
- output type
- constraints

# 4. Guardrails

## One Line Definition

**Guardrails are rules added to prompts to control model behavior and prevent unsafe or incorrect outputs.**

They ensure the AI behaves **safely and predictably**.

---

## Example 1 — Prevent Hallucination

Prompt

```
Answer the question using the document.
If the answer is not in the document, say "I don't know".
```

This prevents the model from **inventing answers**.

---

## Example 2 — Safety Rule

Prompt

You are a helpful assistant.
Do not generate harmful or illegal instructions.
```

This ensures **safe outputs**.

---

## Example 3 — Limiting Response

Prompt

```

Guardrail = **word limit**

---

## Example 4 — Controlled Output

Prompt

```
Return only valid JSON.
Do not include explanations.
```

This prevents extra text.

`````python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate

# Load env variables
load_dotenv()

# Create LLM
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    groq_api_key=os.getenv("GROQ_API_KEY")
)

# Create prompt template
template = PromptTemplate.from_template(
    "Explain {topic} in {points} bullet points."
)

# Fill template
prompt = template.format(
    topic="REST API",
    points=3
)

# Call LLM
response = llm.invoke(prompt)

print(response.content)
`````
